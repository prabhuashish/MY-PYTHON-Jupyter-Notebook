### ==> 3. Data Transformation
***

**Create occupancy percentage column**

```python
df_agg_bookings.head(3)
```

```python
df_agg_bookings["occ_ppt"] = df_agg_bookings["successful_bookings"]/df_agg_bookings["capacity]
```

```python
df_agg_bookings["occ_ppt"] = df_agg_bookings["occ_ppt"].apply(lambda x :round(x*100,2))
df_agg_bookings.head(4)
```

*profit calculation*

```python
df_bookings["profits"] = df_bookings["revenue_generated"] - df_bookings["revenue_realized"]
df_bookings.head(2)
```
