```python
df_raw = pd.read_csv(r"C:\Users\prabh\Code_Basic\64101194a2364\source-code\2_pandas_basics_code\2_dataframe_basics\Exercise\bengaluru_house_prices.csv")
df_raw.head() 
```

```python
df = df_raw[['total_sqft', 'price']]
df.head()
```

```python
df['price_lakhs'] = df['price']
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(df[['total_sqft']], df['price_lakhs'])
```

```python
model.predict([[1500],[1900],[432]])
```

```python
model.coef_ , model.intercept_
```

```python
Now try in Y = mx + b. Y - price_lakhs , M slope coef_ ,  B - intercept , X - Total_sqft.
```

```python
derive_price_lakhs = model.coef_[0]*1500 + (model.intercept_)
derive_price_lakhs
```

```python
derive_price_lakhs = model.coef_[0]*432 + (model.intercept_)
derive_price_lakhs
```

```python
df_raw['BHK'] = df_raw['size'].str.extract(r'(\d+)').astype(int)
df_raw.head()
```

#### From size extracted the BHK, Bedroom into BHK number 

```python
df_norm = df_raw[['total_sqft', 'price','BHK']]
df_norm.head()
```

```python
model = LinearRegression()
model.fit(df_norm[['total_sqft','BHK']], df_norm['price'])
```

```python
model.predict([[1800,3]])
```

```python
model.predict([[2300,4]])
```

```python
price_in_lakhs = model.predict([[2003, 2]])[0]
price_in_lakhs
```

```python
model.coef_ , model.intercept_
```
